```python
!pip install -q beautifulsoup4 requests
import requests
from bs4 import BeautifulSoup
```

```python
url = ("https://store.steampowered.com/search/?specials=1&cc=br&l=portuguese")
pagina = requests.get(url)
print("Status:", pagina.status_code)
```

```python
soup = BeautifulSoup(pagina.text,"html.parser")
```

```python
produtos = soup.select("a.search_result_row")
```

```python
for produto in produtos[:40]:nome = produto.select_one("span.title").get_text(strip=True)
preco_original = produto.select_one(".discount_original_price").get_text(strip=True)

preco_promocional = produto.select_one(".discount_final_price").get_text(strip=True)
desconto = produto.select_one(".discount_pct").get_text(strip=True)

print("Nome:", nome)
print("Preço original:", preco_original)
print("Desconto:", desconto)
print("Preço promocional:", preco_promocional)
print("-" * 50)
```

```python
# Instalando o Selenium
!pip install -q google-colab-selenium
# Importações
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from time import sleep
```

```python
# Configurando o navegador
navegador = gs.Chrome()
# Endereço das cotações do BCE
url = ("https://www.ecb.europa.eu/stats/policy_and_exchange_rates/euro_reference_exchange_rates/html/index.en.html")
# Acessando o endereço do BCE
navegador.get(url)
sleep(2)
print("Página:", navegador.title)

# Localizando os botões de recusa
botoes = navegador.find_elements(
By.XPATH,
"//*[@id='cookieConsent']//button"
"[contains(., 'I do not accept')]"
)

# Clicando no botão visível
for botao in botoes:
  if botao.is_displayed():navegador.execute_script(
  "arguments[0].click();", botao
  )
  break
sleep(1)

# Cotação do dólar (USD)
dolar = navegador.find_element(
By.XPATH,
"//tr[td[normalize-space()='USD']]/td[3]"
).text
# Cotação do real (BRL)
real = navegador.find_element(
By.XPATH,
"//tr[td[normalize-space()='BRL']]/td[3]"
).text

# Exibindo as cotações
print("\nCOTAÇÕES:")
print("1 EUR =", dolar, "USD")
print("1 EUR =", real, "BRL")

# Conversão dólar -> real
dolar_real = float(real) / float(dolar)
print("1 USD =", round(dolar_real, 2), "BRL")

# Clicando no link do dólar
link = navegador.find_element(
By.XPATH,
"//tr[td[normalize-space()='USD']]//a"
)
link.click()
sleep(2)

# Página após o clique
print("\nPágina após o clique:")
print(navegador.title)
print(navegador.current_url)

# Finalizando o navegador
navegador.quit()
```
