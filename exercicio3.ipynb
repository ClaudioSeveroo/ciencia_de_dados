```python
import json

# Nome do arquivo que será utilizado para salvar os dados
NOME_ARQUIVO = "personagem.json"

def criar_e_salvar_personagem():
    """Recebe os dados do usuário e sobrescreve o arquivo com os novos dados."""
    print("=== CRIAÇÃO DE PERSONAGEM ===")
    nome = input("Digite o nome do personagem: ")
    raca = input("Digite a raça (ex: Humano, Elfo, Orc): ")
    classe = input("Digite a classe (ex: Mago, Guerreiro, Ladino): ")
    nivel = input("Digite o nível do personagem: ")

    # organizando os dados em um dicionário
    dados_personagem = {
        "Nome": nome,
        "Raça": raca,
        "Classe": classe,
        "Nível": nivel
    }

    # gravando no arquivo. O modo 'w' garante que o arquivo seja sobrescrito
    with open(NOME_ARQUIVO, 'w', encoding='utf-8') as arquivo:
        json.dump(dados_personagem, arquivo, ensure_ascii=False, indent=4)

    print("\n[!] Personagem gravado com sucesso no arquivo '{}'!\n".format(NOME_ARQUIVO))

def ler_e_apresentar_personagem():
    """Lê o arquivo salvo e imprime os dados na tela."""
    print("=== DADOS DO PERSONAGEM SALVO ===")
    try:
        # abrindo o arquivo no modo de leitura 'r'
        with open(NOME_ARQUIVO, 'r', encoding='utf-8') as arquivo:
            dados = json.load(arquivo)

            # percorrendo o dicionário
            for caracteristica, valor in dados.items():
                print(f"{caracteristica}: {valor}")

    except FileNotFoundError:
        print("Erro: O arquivo de personagem ainda não foi criado.")

# Função principal
def main():
    # 1. Pede os dados e salva (sobrescrevendo se já existir)
    criar_e_salvar_personagem()

    # 2. Chama a função para ler e apresentar o que foi salvo
    ler_e_apresentar_personagem()

# --- Execução principal do programa ---
main()
```
