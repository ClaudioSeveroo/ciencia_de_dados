```python
import numpy as np

# Dados iniciais
dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [98, 11700, 82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200, 95, 120.00, 2]
])

dias = np.array([
    "Segunda", "Terça", "Quarta", "Quinta", "Sexta", "Sábado", "Domingo"
])
```

```python
print("--- Parte 1 ---")
print("1.1 Linhas:", dados.shape[0])
print("1.1 Colunas:", dados.shape[1])
print("1.1 Total de valores:", dados.size)
print("1.1 Dimensões:", dados.ndim)
print("1.1 Tipo dos dados:", dados.dtype)
print("\n")
```

```python
print("--- Parte 2 ---")
print("2.1 Dados de sexta-feira:", dados[4])
print("2.2 Somente faturamento:", dados[:, 1])
print("2.3 Somente reclamações:", dados[:, 4])
print("2.4 Quinta até domingo:\n", dados[3:])
print("2.5 Pedidos, faturamento e clientes:\n", dados[:, 0:3])
print("\n")
```

```python
print("--- Parte 3 ---")
print("3.1 Total de pedidos:", np.sum(dados[:, 0]))
print("3.2 Faturamento total:", np.sum(dados[:, 1]))
print("3.3 Total de clientes:", np.sum(dados[:, 2]))
print("3.4 Média de pedidos:", np.mean(dados[:, 0]))
print("3.5 Média de faturamento:", np.mean(dados[:, 1]))
print("3.6 Média de reclamações:", np.mean(dados[:, 4]))
print("\n")
```

```python
print("--- Parte 4 ---")
pos_maior_pedido = np.argmax(dados[:, 0])
pos_menor_pedido = np.argmin(dados[:, 0])
print("4.1 Maior número de pedidos:", dias[pos_maior_pedido])
print("4.1 Menor número de pedidos:", dias[pos_menor_pedido])

pos_maior_fat = np.argmax(dados[:, 1])
pos_menor_fat = np.argmin(dados[:, 1])
print("4.2 Maior faturamento:", dias[pos_maior_fat])
print("4.2 Menor faturamento:", dias[pos_menor_fat])

amplitude = np.max(dados[:, 1]) - np.min(dados[:, 1])
print("4.3 Amplitude do faturamento:", amplitude)
print("\n")
```

```python
print("--- Parte 5 ---")
# Usei os valores 20000, 150 e 5 para completar os espaços em branco do exercício
mascara_fat = dados[:, 1] > 20000
print("5.1 Faturamento superior a 20.000:", dias[mascara_fat])

mascara_pedidos = dados[:, 0] > 150
print("5.2 Mais de 150 pedidos:", dias[mascara_pedidos])

mascara_reclamacoes = dados[:, 4] > 5
print("5.3 Mais de 5 reclamações:", dias[mascara_reclamacoes])

print("5.4 Faturamentos com mais de 150 pedidos:", dados[mascara_pedidos, 1])

media_semana = np.mean(dados[:, 1])
mascara_abaixo_media = dados[:, 1] < media_semana
print("5.5 Dias com faturamento abaixo da média:", np.sum(mascara_abaixo_media))
print("\n")
```

```python
print("--- Parte 6 ---")
classificacao = np.where(dados[:, 1] >= 20000, "Alto desempenho", "Desempenho normal")
print("6.1 Classificação dos dias:\n", classificacao)

dias_alto_desempenho = np.sum(classificacao == "Alto desempenho")
print("6.2 Quantos dias de alto desempenho:", dias_alto_desempenho)
print("\n")
```

```python
print("--- Parte 7 ---")
fat = dados[:, 1]
print("Média:", np.mean(fat))
print("Mediana:", np.median(fat))
print("Desvio padrão:", np.std(fat))
print("Mínimo:", np.min(fat))
print("Máximo:", np.max(fat))
print("O faturamento apresenta grande variação? Sim, o desvio padrão é muito alto (mais de 4500) e a diferença entre o mínimo (11700) e o máximo (24500) é enorme.")
print("\n")
```

```python
print("--- Parte 8 ---")
print("8.1 Média por indicador:", np.mean(dados, axis=0))
print("8.2 Maior valor por indicador:", np.max(dados, axis=0))
print("8.3 Menor valor por indicador:", np.min(dados, axis=0))
print("8.4 Comparando os resultados, as Reclamações têm a maior variação relativa, porque elas vão de 2 até 9 (cresce mais de 4 vezes), enquanto os outros números não sobem tanto assim proporcionalmente.")
print("\n")
```

```python
print("--- Parte 9 ---")
dia_melhor_fat = dias[np.argmax(dados[:, 1])]
print("9.1 Melhor dia em faturamento:", dia_melhor_fat)

dia_mais_pedidos = dias[np.argmax(dados[:, 0])]
print("9.2 Tem a maior quantidade de pedidos?")
if dia_melhor_fat == dia_mais_pedidos:
    print("Sim")
else:
    print("Não")

dia_mais_reclamacoes = dias[np.argmax(dados[:, 4])]
print("9.3 Tem o maior número de reclamações?")
if dia_melhor_fat == dia_mais_reclamacoes:
    print("Sim")
else:
    print("Não")

filtro_dois = (dados[:, 0] > 150) & (dados[:, 1] > 20000)
print("9.4 Dias com +150 pedidos e fat > 20000:", np.sum(filtro_dois))
print("\n")
```

```python
print("--- Parte 10 ---")
print("RESUMO DA SEMANA")
print("Total de pedidos:", np.sum(dados[:, 0]))
print("Faturamento total:", np.sum(dados[:, 1]))
print("Média diária de pedidos:", np.mean(dados[:, 0]))
print("Média diária de faturamento:", np.mean(dados[:, 1]))
print("Melhor dia:", dias[np.argmax(dados[:, 1])])
print("Pior dia:", dias[np.argmin(dados[:, 1])])
print("Maior número de reclamações:", np.max(dados[:, 4]))
print("Dias de alto desempenho:", np.sum(dados[:, 1] >= 20000))
```
