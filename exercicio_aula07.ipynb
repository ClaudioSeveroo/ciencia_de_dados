```python
import math

# Criação da classe
class Calc:

    def somar(self, a: float, b: float) -> float:
        return a + b

    def subtrair(self, a: float, b: float) -> float:
        return a - b

    def multiplicar(self, a: float, b: float) -> float:
        return a * b

    def dividir(self, a: float, b: float) -> float:
        if b == 0:
            return "não tem como dividir por 0"
        return a / b

    def raiz_quadrada(self, numero: float) -> float:
        if numero < 0:
            raise ValueError("insira números positivos")
        return math.sqrt(numero)

```

```python
# Instancia a calculadora
calculadora = Calc()

loop = True
while loop:
    opcao = int(input("Qual operação voce quer fazer? 1 = +, 2 = -, 3 = *, 4 = /, 5 = raiz, 0 = Sair: "))

    match opcao:
        case 0:
            print("saindo")
            loop = False
        case 1:
            a = float(input("digite um número: "))
            b = float(input("digite outro número: "))
            print("resultado:", calculadora.somar(a, b))
        case 2:
            a = float(input("digite um número: "))
            b = float(input("digite outro número: "))
            print("resultado:", calculadora.subtrair(a, b))
        case 3:
            a = float(input("digite um número: "))
            b = float(input("digite outro número: "))
            print("resultado:", calculadora.multiplicar(a, b))
        case 4:
            a = float(input("digite um número: "))
            b = float(input("digite outro número: "))
            print("resultado:", calculadora.dividir(a, b))
        case 5:
            numero = float(input("digite um número: "))
            print("resultado:", calculadora.raiz_quadrada(numero))
        case _:
            print("opção inválida")
```
