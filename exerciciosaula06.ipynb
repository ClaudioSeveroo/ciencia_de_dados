```python
#Escreva um programa em Python que gere um número aleatório de 0 a 15, usando a biblioteca 'random'
#e o método 'randint()'. Na sequência, utilizando um loop faça com que o usuário tente acertar o número gerado.
#A cada vez que o número é acertado, armazene num arquivo o número gerado e a quantidade de
#tentativas que o usuário realizou para acertar. Quando o usuário decidir parar de adivinhar, mostre todo o
# histórico que ele construiu durante a execução do programa.

import random
numeroAleatorio = random.randint(1,3)
loop = True
cont = 1
# criando arquivo
arquivo = open("/content/meu_arquivo.txt","w")
while(loop):
  numero = int (input ("Informe um numero: "))
  if (numero == numeroAleatorio):
    print("show de bola, acertou o número.")
    print(f"Tentativas = {cont}")
    sair = (input ("Informe se quer sair (S = sim N = não): "))
    # alterando arquivo
    arquivo.write(f"Número: {numeroAleatorio}\n")
    arquivo.write(f"Tentativas:{cont}\n-------------\n")
    if (sair == "S" or sair == "s"):
      loop = False
  else:
    print("errado")
    cont = cont + 1
# fechando e salvando arquivo
arquivo.close()

```

```python
#Crie um programa que permita adicionar itens para uma lista de compras.
#O usuário vai adicionando a quantidade e o valor de cada item da lista.
#Ao final, deve mostrar o valor total das compras e salvar a lista num arquivo .CSV.

print("--- Sistema de Lista de Compras ---")

compras = []
total_compras = 0.0

# Loop para ler os itens
while True:
    nome = input("Digite o nome do produto (ou 'sair' para terminar): ")

    if nome.lower() == 'sair':
        break

    quantidade = float(input("Quantidade: "))
    valor = float(input("Valor unitário: R$ "))

    subtotal = quantidade * valor
    total_compras += subtotal

    # guarda os dados como uma lista dentro de outra lista
    compras.append([nome, quantidade, valor, subtotal])
    print(f"Subtotal do {nome}: R$ {subtotal:.2f}\n")

print("\n--- Resumo da sua compra ---")
for item in compras:
    print(f"Produto: {item[0]} | Qtd: {item[1]} | Unitário: R$ {item[2]:.2f} | Total: R$ {item[3]:.2f}")

print(f"Valor Total de todas as compras: R$ {total_compras:.2f}")

# salva no arquivo CSV
arquivo = open("lista_compras.csv", "w", encoding="utf-8")

# Escreve o cabeçalho
arquivo.write("Produto,Quantidade,Valor_Unitario,Subtotal\n")

# escreve cada item linha por linha
for item in compras:
    arquivo.write(f"{item[0]},{item[1]},{item[2]},{item[3]}\n")

arquivo.close()

print("\nArquivo 'lista_compras.csv' salvo com sucesso!")
```
