```python
# uso da biblioteca pandas
import pandas as pd
# nomes dos campos
nomes = ['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth', 'Class']
df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data', names = nomes)
print(f'\n {df}')
```

```python
# Separando o dataset
# X será a matriz de características
X = df[df.columns.difference(['Class'])].values
print (f'X: {X}')

# y será o vetor com a classificação ou algo (target)
y = df['Class'].values
print (f'Y: {y}')
```

```python
# importamos a biblioteca sklearn que possui a implementação do algortimo
from sklearn.neighbors import KNeighborsClassifier
# definimos o numero de vizinhos = 3 para classificação
flor_iris_classificador = KNeighborsClassifier(n_neighbors = 3)
# uso do método fit para treinar o modelo - passando os valores do dataset
flor_iris_classificador.fit(X, y)
```

```python
# Desempenho do KNN
# Usando validação cruzada para
from sklearn.model_selection import cross_val_score
# quebra o dataset em 5 partes e usa uma delas para teste (validação)
scores_dt = cross_val_score(flor_iris_classificador, X, y, scoring='accuracy', cv=5)
# o testes é aletório, pegamos os valores médios
print(scores_dt.mean())
```

```python
# Treinando e avaliando - alterando o numero de vizinhos
for i in range(1,11):
    flor_iris_classificador = KNeighborsClassifier(n_neighbors = i)
    flor_iris_classificador.fit(X, y)
    scores_dt = cross_val_score(flor_iris_classificador, X, y, scoring='accuracy', cv=5)
    print(f'Vizinhos: {i}  Acurácia: {scores_dt.mean()}')
```

```python
# Avaliando a classificação com novos dados
novos_exemplos = [[5.2,2.4,7.0,3.2], [4.2,1.2,5.8,2.7], [1.6,0.5,5.0,3.6]]
# Avaliando a classificação
print(flor_iris_classificador.predict(novos_exemplos))
```

```python
# Testando a classificação de forma interativa
sepal_length = input('Comprimento da sépala: ')
sepal_width = input('Largura da sépala   : ')
petal_length = input('Comprimento da Pétala: ')
petal_width = input('Largura da Pétala   : ')
teste = [[float(sepal_length), float(sepal_width), float(petal_length), float(petal_width)]]
print(f'Amostra Classificada como: {flor_iris_classificador.predict(teste)}')
```
