```python
# uso da biblioteca pandas
import pandas as pd
# nomes dos campos
nomes = ['fixed acidity','volatile acidity','citric acid','residual sugar','chlorides','free sulfur dioxide','total sulfur dioxide','density','pH','sulphates','alcohol','quality']
df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv', names = nomes, sep=';', skiprows=1)
print(f'\n {df}')
```

```python
def classificar(nota):
    if nota <= 4:
        return 'Baixa'
    elif nota <= 6:
        return 'Média'
    else:
        return 'Alta'

df['Classificacao'] = df['quality'].apply(classificar)

X = df[df.columns.difference(['quality', 'Classificacao'])].values
y = df['Classificacao'].values

from sklearn.neighbors import KNeighborsClassifier
vinho_classificador = KNeighborsClassifier(n_neighbors = 5)
vinho_classificador.fit(X, y)

from sklearn.model_selection import cross_val_score
scores_dt = cross_val_score(vinho_classificador, X, y, scoring='accuracy', cv=5)
print(f'\nAcurácia média com 5 vizinhos: {scores_dt.mean()}')

print('\n--- Avaliando variação no número de vizinhos ---')
for i in range(1, 11):
    vinho_classificador = KNeighborsClassifier(n_neighbors = i)
    vinho_classificador.fit(X, y)
    scores_dt = cross_val_score(vinho_classificador, X, y, scoring='accuracy', cv=5)
    print(f'Vizinhos: {i}  Acurácia: {scores_dt.mean()}')
```
